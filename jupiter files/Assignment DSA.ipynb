### Part A - Q2: Recursive Factorial, Sum of n Natural Numbers, and Power

```python
def fact(n):
    if n == 0 or n == 1:
        return 1
    return n * fact(n - 1)

def sum_n(n):
    if n <= 0:
        return 0
    return n + sum_n(n - 1)

def power(x, p):
    if p == 0:
        return 1
    return x * power(x, p - 1)

print("fact(5) =", fact(5))
print("sum_n(10) =", sum_n(10))
print("power(2, 6) =", power(2, 6))
```

### Part A - Q3: Recursive Reverse in Place

```python
def rev_seq(arr, start, stop):
    if start < stop:
        arr[start], arr[stop] = arr[stop], arr[start]
        rev_seq(arr, start + 1, stop - 1)

a = [4, 7, 1, 9, 3, 8]
print("before:", a)
rev_seq(a, 0, len(a) - 1)
print("after:", a)
```

### Part A - Q4: Recursive Check if Sorted

```python
def is_sorted(arr, i=0):
    if i >= len(arr) - 1:
        return True
    if arr[i] > arr[i + 1]:
        return False
    return is_sorted(arr, i + 1)

print(is_sorted([1, 3, 5, 5, 9]))
print(is_sorted([2, 8, 4, 10]))
```

### Part A - Q5: Naive Recursive Fibonacci

```python
calls = 0

def fib(n):
    global calls
    calls += 1
    if n <= 1:
        return n
    return fib(n - 1) + fib(n - 2)

print("fib(5) =", fib(5))
print("total calls for n=5:", calls)
```

### Part A - Q7: Tower of Hanoi

```python
def hanoi(n, src, aux, dst):
    if n == 1:
        print("move disk 1 from", src, "to", dst)
        return
    hanoi(n - 1, src, dst, aux)
    print("move disk", n, "from", src, "to", dst)
    hanoi(n - 1, aux, src, dst)

hanoi(3, "A", "B", "C")
```

### Part A - Q9: Iterative Version of Tail-Recursive Sequence Processing

```python
def rev_iter(arr):
    start = 0
    stop = len(arr) - 1
    while start < stop:
        arr[start], arr[stop] = arr[stop], arr[start]
        start += 1
        stop -= 1

nums = [10, 20, 30, 40, 50]
rev_iter(nums)
print("reversed:", nums)
```

### Part C - Q2: Stack Class Using Python List

```python
class Empty(Exception):
    pass

class Stack:
    def __init__(self):
        self.data = []

    def __len__(self):
        return len(self.data)

    def is_empty(self):
        return len(self.data) == 0

    def push(self, val):
        self.data.append(val)

    def top(self):
        if self.is_empty():
            raise Empty("stack is empty")
        return self.data[-1]

    def pop(self):
        if self.is_empty():
            raise Empty("stack is empty")
        return self.data.pop()

st = Stack()
st.push(10)
st.push(20)
print("top:", st.top())
print("pop:", st.pop())
print("len:", len(st))

try:
    st.pop()
    st.pop()
except Empty as e:
    print("error:", e)
```

### Part C - Q8: Circular-Array Queue

```python
class CircularQueue:
    def __init__(self, cap=5):
        self.data = [None] * cap
        self.size = 0
        self.front = 0

    def is_empty(self):
        return self.size == 0

    def first(self):
        if self.is_empty():
            raise Exception("queue empty")
        return self.data[self.front]

    def enqueue(self, val):
        if self.size == len(self.data):
            raise Exception("queue full")
        avail = (self.front + self.size) % len(self.data)
        self.data[avail] = val
        self.size += 1

    def dequeue(self):
        if self.is_empty():
            raise Exception("queue empty")
        ans = self.data[self.front]
        self.data[self.front] = None
        self.front = (self.front + 1) % len(self.data)
        self.size -= 1
        return ans

q = CircularQueue(4)
q.enqueue(5)
q.enqueue(12)
q.enqueue(19)
print("deq:", q.dequeue())
q.enqueue(25)
q.enqueue(31)
print("first:", q.first())
print("array:", q.data)
```

### Part C - Q10: Circular-Array Deque

```python
class CircularDeque:
    def __init__(self, cap=6):
        self.data = [None] * cap
        self.size = 0
        self.front = 0

    def add_first(self, val):
        if self.size == len(self.data):
            raise Exception("deque full")
        self.front = (self.front - 1) % len(self.data)
        self.data[self.front] = val
        self.size += 1

    def add_last(self, val):
        if self.size == len(self.data):
            raise Exception("deque full")
        avail = (self.front + self.size) % len(self.data)
        self.data[avail] = val
        self.size += 1

    def delete_first(self):
        if self.size == 0:
            raise Exception("deque empty")
        ans = self.data[self.front]
        self.data[self.front] = None
        self.front = (self.front + 1) % len(self.data)
        self.size -= 1
        return ans

    def delete_last(self):
        if self.size == 0:
            raise Exception("deque empty")
        back = (self.front + self.size - 1) % len(self.data)
        ans = self.data[back]
        self.data[back] = None
        self.size -= 1
        return ans

dq = CircularDeque(5)
dq.add_last(10)
dq.add_first(5)
dq.add_last(20)
print("del first:", dq.delete_first())
print("del last:", dq.delete_last())
print("remaining:", dq.data)
```

### Part D - Q3: Singly Linked List Operations

```python
class Node:
    def __init__(self, elem, nxt=None):
        self.elem = elem
        self.nxt = nxt

class SinglyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0

    def insert_first(self, val):
        newest = Node(val, self.head)
        self.head = newest
        if self.size == 0:
            self.tail = newest
        self.size += 1

    def insert_last(self, val):
        newest = Node(val, None)
        if self.size == 0:
            self.head = newest
        else:
            self.tail.nxt = newest
        self.tail = newest
        self.size += 1

    def delete_first(self):
        if self.size == 0:
            raise Exception("list empty")
        val = self.head.elem
        self.head = self.head.nxt
        self.size -= 1
        if self.size == 0:
            self.tail = None
        return val

    def traverse(self):
        cur = self.head
        out = []
        while cur is not None:
            out.append(cur.elem)
            cur = cur.nxt
        return out

sll = SinglyLinkedList()
sll.insert_first(20)
sll.insert_first(10)
sll.insert_last(30)
print("list:", sll.traverse())
print("deleted:", sll.delete_first())
print("after delete:", sll.traverse())
```

### Part D - Q6: Favorites List ADT

```python
class FavoritesList:
    class Item:
        def __init__(self, val):
            self.val = val
            self.count = 0

    def __init__(self):
        self.items = []

    def _find(self, e):
        for i in range(len(self.items)):
            if self.items[i].val == e:
                return i
        return -1

    def access(self, e):
        idx = self._find(e)
        if idx == -1:
            self.items.append(self.Item(e))
            idx = len(self.items) - 1
        self.items[idx].count += 1
        # move forward to keep sorted by count
        while idx > 0 and self.items[idx].count > self.items[idx - 1].count:
            self.items[idx], self.items[idx - 1] = self.items[idx - 1], self.items[idx]
            idx -= 1

    def remove(self, e):
        idx = self._find(e)
        if idx != -1:
            self.items.pop(idx)

    def top(self, k):
        res = []
        for i in range(min(k, len(self.items))):
            res.append((self.items[i].val, self.items[i].count))
        return res

fav = FavoritesList()
for song in ["a", "b", "a", "c", "b", "a"]:
    fav.access(song)
print("top 2:", fav.top(2))
fav.remove("b")
print("after removing b:", fav.top(3))
```
