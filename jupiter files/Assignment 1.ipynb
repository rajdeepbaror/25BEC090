### 3. Strings

```python
s = input("enter a string: ")

print("characters:", len(s))

v = 0
c = 0
for ch in s:
    if ch.isalpha():
        if ch.lower() in "aeiou":
            v += 1
        else:
            c += 1

print("vowels:", v)
print("consonants:", c)

rev = s[::-1]
print("reversed:", rev)

if s.lower() == rev.lower():
    print("palindrome: yes")
else:
    print("palindrome: no")
```

### 4. Sets and Dictionaries

```python
s1 = {1, 2, 3, 4, 5}
s2 = {4, 5, 6, 7, 8}

print("union:", s1.union(s2))
print("intersection:", s1.intersection(s2))
print("difference (s1 - s2):", s1.difference(s2))
print("symmetric difference:", s1.symmetric_difference(s2))

marks = {
    "rahul": 84,
    "aman": 91,
    "sneha": 88,
    "priya": 95,
    "karan": 76
}

topper = ""
high = -1
for name in marks:
    if marks[name] > high:
        high = marks[name]
        topper = name

print("topper:", topper, "with", high, "marks")
```

### 5. GPA Calculator

```python
n = int(input("no. of courses: "))

total_pts = 0
total_cr = 0

for i in range(n):
    gp = float(input(f"course {i+1} grade point: "))
    cr = float(input(f"course {i+1} credits: "))
    total_pts += gp * cr
    total_cr += cr

gpa = total_pts / total_cr
print("gpa =", round(gpa, 2))
```

### 6. Functions and Iteration

```python
def fact(n):
    res = 1
    for i in range(1, n + 1):
        res *= i
    return res

def fib(n):
    a, b = 0, 1
    ans = []
    for _ in range(n):
        ans.append(a)
        a, b = b, a + b
    return ans

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

print("factorial of 5:", fact(5))
print("fibonacci (7 terms):", fib(7))
print("is 13 prime?", is_prime(13))
print("is 12 prime?", is_prime(12))
```

### 10. Mini Programming Problem

```python
students = {}
print("1.add  2.update  3.search  4.delete  5.display  6.exit")

while True:
    ch = input("choice: ")

    if ch == "1":
        name = input("name: ")
        m = float(input("marks: "))
        students[name] = m
        print("added")
    elif ch == "2":
        name = input("name: ")
        if name in students:
            students[name] = float(input("new marks: "))
            print("updated")
        else:
            print("not found")
    elif ch == "3":
        name = input("name: ")
        if name in students:
            print(name, ":", students[name])
        else:
            print("not found")
    elif ch == "4":
        name = input("name: ")
        if name in students:
            del students[name]
            print("deleted")
        else:
            print("not found")
    elif ch == "5":
        print("records:", students)
    elif ch == "6":
        print("bye")
        break
    else:
        print("wrong choice")
```
